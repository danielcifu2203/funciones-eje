# Análisis básico de pH y temperatura del agua
Ejercicio de aprendizaje con datos simulados de 10 días.

```python
# 1. Datos de 10 dias
ph = [7.2, 7.1, 6.9, 6.8, 6.7, 6.6, 6.5, 6.7, 6.9, 7.0]
temperatura = [18.5, 19.0, 20.2, 21.5, 22.0, 23.4, 24.0, 22.8, 21.0, 19.5]

print("Datos de pH:", ph)
print("Datos de temperatura:", temperatura)
```

## 2. Promedio del pH

```python
def promedio_ph():
    promedio = sum(ph) / len(ph)
    print("Promedio del pH:", round(promedio, 2))

promedio_ph()
```

## 3. Promedio de temperatura

```python
def promedio_temperatura():
    promedio = sum(temperatura) / len(temperatura)
    print("Temperatura promedio:", round(promedio, 2), "grados C")

promedio_temperatura()
```

## 4. pH maximo y minimo

```python
def ph_mayor_menor():
    mayor = max(ph)
    menor = min(ph)
    print("pH mayor:", mayor)
    print("pH menor:", menor)

ph_mayor_menor()
```

## 5. Temperatura maxima y minima

```python
def temperatura_mayor_menor():
    mayor = max(temperatura)
    menor = min(temperatura)
    print("Temperatura mayor:", mayor, "grados C")
    print("Temperatura menor:", menor, "grados C")

temperatura_mayor_menor()
```

## 6. Clasificar el pH

```python
def analizar_ph(valor):
    if valor < 7:
        print("El agua tiene pH acido")
    elif valor == 7:
        print("El agua tiene pH neutro")
    else:
        print("El agua tiene pH basico")

analizar_ph(6.5)
```

## 7. Mostrar todos los datos

```python
def mostrar_datos():
    for i in range(10):
        print("Dia:", i + 1)
        print("pH:", ph[i])
        print("Temperatura:", temperatura[i], "grados C")
        print("--------------------")

mostrar_datos()
```

## 8. Diferencia entre pH mayor y menor

```python
def diferencia_ph():
    mayor = max(ph)
    menor = min(ph)
    diferencia = mayor - menor
    print("Diferencia del pH:", round(diferencia, 2))

diferencia_ph()
```

## 9. Diferencia de temperatura

```python
def diferencia_temperatura():
    mayor = max(temperatura)
    menor = min(temperatura)
    diferencia = mayor - menor
    print("Diferencia de temperatura:", round(diferencia, 2), "grados C")

diferencia_temperatura()
```
